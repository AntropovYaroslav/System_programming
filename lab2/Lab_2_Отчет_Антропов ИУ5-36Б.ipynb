## Задание 1. Реверс строки

**Условие:** вывести строку S = `yyblCMJzdDhDeSKlWGmMX` в обратном порядке.

**Исходный код (FASM):**

```asm
format ELF64
public _start
public my_exit

section '.data' writable
    S       db 'yyblCMJzdDhDeSKlWGmMX'
    SLen    = 21
    place   db 1

section '.text' executable
_start:
    mov     rsi, S + SLen - 1
    mov     rcx, SLen
rev_loop:
    mov     al, [rsi]
    mov     byte [place], al
    push    rcx
    push    rsi
    call    print_symb
    pop     rsi
    pop     rcx
    dec     rsi
    loop    rev_loop
    mov     al, 0xA
    mov     byte [place], al
    call    print_symb
    call    my_exit

print_symb:
    mov     eax, 4
    mov     ebx, 1
    mov     ecx, place
    mov     edx, 1
    int     0x80
    ret

section '.my_exit' executable
my_exit:
    mov     eax, 1
    mov     ebx, 0
    int     0x80
```

**Команды сборки и запуска:**

```bash
fasm task1.asm task1.o
ld task1.o -o task1
./task1
```

## Задание 2. Матрица M × K из символа `:`

**Условие:** заполнить память N = 105 символами `:`, вывести в виде матрицы 7 × 15 (7 символов в строке, 15 строк).

**Исходный код (FASM, Linux x86-64):**

```asm
; task2.asm — матрица 7×15 из символа ':'
format ELF64

public _start
public my_exit

M       = 7
K       = 15
N       = M * K
D       = ':'

section '.bss' writable
    buf     db N dup(D)
    place   db 1

section '.text' executable
_start:
    mov     rsi, buf
    mov     rcx, K
row_loop:
    push    rcx
    mov     rcx, M
col_loop:
    mov     al, [rsi]
    mov     byte [place], al
    push    rcx
    push    rsi
    call    print_symb
    pop     rsi
    pop     rcx
    inc     rsi
    loop    col_loop
    mov     al, 0xA
    mov     byte [place], al
    call    print_symb
    pop     rcx
    loop    row_loop
    call    my_exit

print_symb:
    mov     eax, 4
    mov     ebx, 1
    mov     ecx, place
    mov     edx, 1
    int     0x80
    ret

section '.my_exit' executable
my_exit:
    mov     eax, 1
    mov     ebx, 0
    int     0x80
```

**Команды сборки и запуска:**

```bash
fasm task2.asm task2.o
ld task2.o -o task2
./task2
```

## Задание 3. Треугольник из 105 символов `:`

**Условие:** вывести 105 символов `:` в виде треугольника. Так как 105 = 1+2+3+...+14, треугольник состоит из 14 строк.

**Исходный код (FASM):**

```asm
; task3.asm — треугольник из символа ':' (105 символов)
format ELF64

public _start
public my_exit

ROWS    = 14
CHAR    = ':'

section '.data' writable
    place   db 1

section '.text' executable
_start:
    mov     bl, 1
line_loop:
    cmp     bl, ROWS
    jg      done
    mov     cl, bl
ch_loop:
    mov     dl, CHAR
    mov     byte [place], dl
    push    rcx
    call    print_symb
    pop     rcx
    dec     cl
    jnz     ch_loop
    mov     dl, 0xA
    mov     byte [place], dl
    call    print_symb
    inc     bl
    jmp     line_loop
done:
    call    my_exit

print_symb:
    push    rax
    push    rbx
    push    rcx
    push    rdx
    mov     eax, 4
    mov     ebx, 1
    mov     ecx, place
    mov     edx, 1
    int     0x80
    pop     rdx
    pop     rcx
    pop     rbx
    pop     rax
    ret

section '.my_exit' executable
my_exit:
    mov     eax, 1
    mov     ebx, 0
    int     0x80
```

**Команды сборки и запуска:**

```bash
fasm task3.asm task3.o
ld task3.o -o task3
./task3
```

**Результат работы:**

![Скриншот task3](img/task3.png)

Вывод: 14 строк, каждая следующая на 1 символ длиннее предыдущей.

## Задание 4. Сумма цифр числа N = 5447175926

**Условие:** найти сумму цифр числа N.

**Проверка вручную:** 5+4+4+7+1+7+5+9+2+6 = **50**.

**Исходный код (FASM):**

```asm
; task4.asm — сумма цифр числа N=5447175926
format ELF64

public _start
public my_exit

section '.data' writable
    num     db '5447175926'
    nlen    = 10
    place   db 1

section '.text' executable
_start:
    mov     rsi, num
    mov     rcx, nlen
    xor     rbx, rbx
sum_loop:
    movzx   ax, byte [rsi]
    sub     ax, '0'
    add     bx, ax
    inc     rsi
    loop    sum_loop
    mov     ax, bx
    xor     dx, dx
    mov     cx, 10
    div     cx
    push    dx
    add     al, '0'
    mov     byte [place], al
    call    print_symb
    pop     dx
    add     dl, '0'
    mov     byte [place], dl
    call    print_symb
    mov     al, 0xA
    mov     byte [place], al
    call    print_symb
    call    my_exit

print_symb:
    push    rax
    push    rbx
    push    rcx
    push    rdx
    mov     eax, 4
    mov     ebx, 1
    mov     ecx, place
    mov     edx, 1
    int     0x80
    pop     rdx
    pop     rcx
    pop     rbx
    pop     rax
    ret

section '.my_exit' executable
my_exit:
    mov     eax, 1
    mov     ebx, 0
    int     0x80
```

**Команды сборки и запуска:**

```bash
fasm task4.asm task4.o
ld task4.o -o task4
./task4
```

**Результат работы:**

![Скриншот task4](img/task4.png)

Вывод: `50`

## Задание 5. Сумма цифр числа N на языке C

**Условие:** реализовать задание 4 на языке C. Сравнить размеры исполняемых файлов. Постараться сократить размер на ASM и на C.

**Исходный код (C):**

```c
#include <stdio.h>

int main(void) {
    char num[] = "5447175926";
    int sum = 0;
    for (int i = 0; num[i]; i++)
        sum += num[i] - '0';
    printf("%d\n", sum);
    return 0;
}
```

**Команды сборки и запуска:**

```bash
gcc -O2 task5.c -o task5_O2
gcc -Os -s task5.c -o task5_Os
./task5_O2
./task5_Os
```
